### Анализ типа Milestone

```python
import json
ms = df[df['fact_type'] == 'milestone']
ms.head()
```

```python
ms_claims = pd.json_normalize(ms['claim'].apply(json.loads))
ms_claims = ms_claims[ ['stat', 'scope', 'threshold', 'subject_id', 'current_value'] ]
ms_claims.head()
```

```python
ms_claims['stat'].value_counts(normalize=True)
```
